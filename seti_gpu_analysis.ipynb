```python
from turbo_seti.find_doppler.find_doppler import FindDoppler
import os

# Define a list of datasets you want to process sequentially
# (You can add more file dictionaries/URLs here as you find them)
datasets = [
    {
        "file_name": "Voyager1.single_coarse.fine_res.h5",
        "target_url": "http://blpd0.ssl.berkeley.edu/Voyager_data/Voyager1.single_coarse.fine_res.h5"
    },
    # Example template for adding a future file:
    # {
    #     "file_name": "another_file.h5",
    #     "target_url": "http://blpd0.ssl.berkeley.edu/path/to/another_file.h5"
    # }
]

# Ensure output directory exists
output_dir = "./output"
os.makedirs(output_dir, exist_ok=True)

# Loop through each dataset sequentially
for item in datasets:
    f_name = item["file_name"]
    t_url = item["target_url"]

    print(f"\n==============================================")
    print(f"[+] Processing dataset: {f_name}")
    print(f"==============================================")

    # 1. Download if not already present locally
    if not os.path.exists(f_name):
        print(f"[+] Downloading from {t_url}...")
        !wget -nc {t_url}
        print(f"[✓] Download of {f_name} complete!")
    else:
        print(f"[✓] {f_name} already exists locally. Skipping download.")

    # 2. Run GPU-accelerated de-doppler search
    print(f"[+] Initializing GPU pipeline for {f_name}...")
    fdop = FindDoppler(
        datafile=f_name,
        max_drift=4.0,
        snr=10.0,
        out_dir=output_dir,
        gpu_backend=True
    )

    print(f"[+] Launching search... (Results will save to {output_dir}/)")
    fdop.search()
    print(f"[✓] Completed analysis for {f_name}!")

print("\nAll queued datasets have been successfully processed!")
```

```python
from turbo_seti.find_doppler.find_doppler import FindDoppler
import os
from google.colab import files

# Curated list of verified, real-space dataset files
dataset_queue = [
    {
        "name": "Voyager_1_Calibration",
        "file_name": "Voyager1.single_coarse.fine_res.h5",
        "url": "http://blpd0.ssl.berkeley.edu/Voyager_data/Voyager1.single_coarse.fine_res.h5"
    },
    {
        "name": "GBT_Waterhole_Sample",
        "file_name": "waterhole_sample.h5",
        "url": "https://huggingface.co/datasets/phanerozoic/breakthrough-listen-gbt-waterhole/resolve/main/data/waterhole_sample.h5"
    }
]

output_dir = "./output"
os.makedirs(output_dir, exist_ok=True)

print(f"[*] Starting automated multi-target queue. Total files: {len(dataset_queue)}")

for index, target in enumerate(dataset_queue, start=1):
    f_name = target["file_name"]
    t_url = target["url"]

    print("\n" + "="*50)
    print(f"[*] Processing Target {index}/{len(dataset_queue)}: {target['name']}")
    print(f"[*] File: {f_name}")
    print("="*50)

    # 1. Download if not already cached
    if not os.path.exists(f_name):
        print(f"[+] Downloading {f_name}...")
        !wget -O {f_name} {t_url}
        print("[✓] Download complete!")
    else:
        print(f"[✓] {f_name} already exists locally. Skipping download.")

    # 2. Run GPU-accelerated pipeline
    print("[+] Launching GPU-accelerated de-doppler search...")
    try:
        fdop = FindDoppler(
            datafile=f_name,
            max_drift=4.0,
            snr=10.0,
            out_dir=output_dir,
            gpu_backend=True
        )
        fdop.search()
        print(f"[✓] Search completed successfully for {target['name']}!")
    except Exception as e:
        print(f"[-] Error during analysis of {f_name}: {e}")
        continue

    # 3. Clean up the heavy .h5 file to save Colab space
    if os.path.exists(f_name):
        os.remove(f_name)
        print(f"[*] Cleared local cache for {f_name}.")

# 4. Automatically trigger downloads of all generated candidate CSVs to your phone
print("\n[+] Pulling all generated CSV reports down to your device...")
for csv_file in os.listdir(output_dir):
    if csv_file.endswith('.csv'):
        csv_path = os.path.join(output_dir, csv_file)
        print(f"[+] Downloading report: {csv_file}")
        files.download(csv_path)

print("\n[🎉] Queue execution fully completed!")
```
